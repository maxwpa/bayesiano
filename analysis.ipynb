## 2️⃣ ESPECIFICAÇÃO TEÓRICA DO MODELO DE AFC

## 3️⃣ AFC ROBUSTA (MLR)

```r
fitMeasures(fit_emocional, c(
  "chisq","df","pvalue","cfi","tli","rmsea","rmsea.ci.lower","rmsea.ci.upper","srmr"
))
```

```r
fit_trabalho <- cfa(
  modelo_trabalho,
  data = dados,
  estimator = "MLR",
  std.lv = TRUE,
  missing = "fiml"
)
```

```r
fitMeasures(fit_trabalho, c(
  "chisq","df","pvalue","cfi","tli","rmsea","rmsea.ci.lower","rmsea.ci.upper","srmr"
))
```

## 8️⃣ INVARIÂNCIA DE MEDIDA

#### 🔹Gênero

```r
fit_bayes_config_gender <- bcfa(
  modelo_emocional,
  data = dados,
  group = "Q1",        # exemplo: gênero
  burnin = 2000,
  sample = 5000,
  n.chains = 4
)
```

```r
summary(fit_bayes_config_gender)
```

```r
fit_bayes_metric_gender <- bcfa(
  modelo_emocional,
  data = dados,
  group = "Q1",
  group.equal = "loadings",
  burnin = 2000,
  sample = 5000,
  n.chains = 4
)
```

```r
summary(fit_bayes_metric_gender)
```

```r
fit_bayes_gender <- bcfa(
  modelo_emocional,
  data = dados,
  group = "Q1",
  burnin = 2000,
  sample = 5000,
  n.chains = 4,
  group.equal = c("loadings", "intercepts")
)
```

```r
summary(fit_bayes_gender)
```

Foi conduzida uma análise de invariância de medida Bayesiana entre homens (n = 150) e mulheres (n = 173). O modelo configural apresentou adequada representação da estrutura fatorial em ambos os grupos, com cargas fatoriais elevadas e intervalos de credibilidade que não incluíram zero. Em seguida, o modelo métrico, no qual as cargas fatoriais foram igualadas entre os grupos, apresentou melhora no ajuste Bayesiano, conforme indicado pelo aumento do logaritmo da verossimilhança marginal. Por fim, o modelo escalar, com cargas e interceptos restritos à igualdade, manteve estabilidade das estimativas e apresentou o melhor ajuste relativo. Esses resultados fornecem evidência robusta de invariância configural, métrica e escalar do instrumento entre os gêneros, permitindo a comparação válida de médias latentes.

## 🔟 COMPARAÇÃO ENTRE FATORES LATENTES

```r
fatores <- c(
  "Consciencia_Emocional",
  "Regulacao_Emocional",
  "Consciencia_Social",
  "Autocontrole_Emocional",
  "Criatividade_Emocional"
)
```

```r
comb <- combn(fatores, 2, simplify = FALSE)
```

```r
contraste_fatores <- function(post, f1, f2) {
  delta <- post[[1]][, paste0(f1, "~1.g2")] -
           post[[1]][, paste0(f2, "~1.g2")]
  
  c(
    Media = mean(delta),
    IC2.5 = quantile(delta, .025),
    IC97.5 = quantile(delta, .975)
  )
}
```

```r
postG <- blavInspect(fit_bayes_gender, "mcmc")

resultados_genero <- do.call(
  rbind,
  lapply(comb, function(x) contraste_fatores(postG, x[1], x[2]))
)

rownames(resultados_genero) <- sapply(comb, paste, collapse = " - ")
resultados_genero
```

```r
postL <- blavInspect(fit_bayes_leader, "mcmc")

resultados_lideranca <- do.call(
  rbind,
  lapply(comb, function(x) contraste_fatores(postL, x[1], x[2]))
)

rownames(resultados_lideranca) <- sapply(comb, paste, collapse = " - ")
resultados_lideranca
```

```r
diff_diff_fatores <- function(postG, postL, f1, f2) {
  
  # diferença entre fatores – mulheres
  delta_mulheres <- postG[[1]][, paste0(f1, "~1.g2")] -
                    postG[[1]][, paste0(f2, "~1.g2")]
  
  # diferença entre fatores – líderes
  delta_lideres <- postL[[1]][, paste0(f1, "~1.g2")] -
                   postL[[1]][, paste0(f2, "~1.g2")]
  
  # diferença das diferenças
  diff_diff <- delta_mulheres - delta_lideres
  
  c(
    Media = mean(diff_diff),
    IC2.5 = quantile(diff_diff, .025),
    IC97.5 = quantile(diff_diff, .975)
  )
}
```

```r
fatores <- c(
  "Consciencia_Emocional",
  "Regulacao_Emocional",
  "Consciencia_Social",
  "Autocontrole_Emocional",
  "Criatividade_Emocional"
)

comb <- combn(fatores, 2, simplify = FALSE)
```

```r
postG <- blavInspect(fit_bayes_gender, "mcmc")
postL <- blavInspect(fit_bayes_leader, "mcmc")

resultados_diffdiff <- do.call(
  rbind,
  lapply(comb, function(x)
    diff_diff_fatores(postG, postL, x[1], x[2]))
)

rownames(resultados_diffdiff) <- sapply(comb, paste, collapse = " − ")

resultados_diffdiff
```

```r
resultados_diffdiff <- as.data.frame(resultados_diffdiff)

resultados_diffdiff$Robusto <-
  !(resultados_diffdiff$IC2.5 < 0 & resultados_diffdiff$IC97.5 > 0)

resultados_diffdiff
```
